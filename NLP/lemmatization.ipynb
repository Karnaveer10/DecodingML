Lemmatization vs Stemming:

Stemming = crude chopping → "studies" → "studi" (may not be real word)

Lemmatization = smart dictionary lookup → "studies" → "study" (valid word)​

Lemma: The actual dictionary base form of a word (e.g., "better" → "good")​

NLTK WordNetLemmatizer:

Wrapper around WordNet corpus (huge English dictionary)

Uses morphy() function to find correct lemma

Context-aware: Needs POS (part-of-speech) for accuracy
Always returns valid English words, unlike stemming's pseudo-stems

removing inflectional endings (such as tense, pluralization, or case markers).
lemmetization takes more time and computation since it has to search a large corpus


```python
from nltk.stem import WordNetLemmatizer
lemm = WordNetLemmatizer()
```

```python
import nltk
nltk.download('wordnet')
```

```python
'''
POS- Noun-n
verb-v
adjective-a
adverb-r
'''
lemm.lemmatize("going",pos='v')
## we have to specify parts of speech
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python

for word in words:
    print(word+"---->"+lemm.lemmatize(word,pos='v'))

## try with various pos , like n etc 
## to find pos we can use pos tagging 
```
