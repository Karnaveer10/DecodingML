```python
# Stemming Definition: Reduces words to their base or root form (stem) by removing prefixes, suffixes, or inflections, like "running," "runs," and "runner" to "run."​

# Key Process: Uses heuristic rules for fast normalization, unlike precise lemmatization.​

# Importance in NLP:

# Reduces vocabulary size, improving efficiency in search, classification, and analysis.​

# Enhances pattern recognition by grouping word variants.​

# Boosts tasks like information retrieval, topic modeling, and model training speed.
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

PorterStemmer: NLTK's classic English stemmer using rule-based suffix removal (e.g., "running" → "run"); accurate but English-only and gentle.​

RegexpStemmer: Customizable NLTK stemmer via regex patterns to strip prefixes/suffixes (e.g., "un-" or "-ly"); flexible but requires manual rules, error-prone without tuning.​

Snowball Stemmer: Enhanced multilingual NLTK stemmer (Porter2); more aggressive and faster than Porter, supports 15+ languages with better accuracy (e.g., "fairly" → "fair").

```python
from nltk.stem import PorterStemmer
stemming=PorterStemmer()
for w in words:
    print(w+"----->"+stemming.stem(w))
    
##Failed for proper nouns like history 
stemming.stem('congratulations')
```

```python
from nltk.stem import RegexpStemmer
reg = RegexpStemmer('ing$|s$|e$|able$', min=4)
## words ending with given suffix , min =4 means dont stem words with length <4
for w in words:
    print(w+"----->"+reg.stem(w))
    
```

```python
from nltk.stem import SnowballStemmer
snow = SnowballStemmer('english')
## has various languages and can ignore stop words 
for w in words:
    print(w+"----->"+snow.stem(w))
    
```

```python
print(stemming.stem("fairly"),stemming.stem("sportingly"))
print(snow.stem("fairly"),snow.stem("sportingly"))
print(snow.stem('goes'))
print(stemming.stem('goes'))


##Stemming is still not accurate , for better accuracy use lemmetization 

```
