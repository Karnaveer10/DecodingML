```python
corpus ='''The quick brown fox jumps over the lazy dog near the river bank.
In the morning, the weather was cold and the sky was filled with clouds.
She is studying natural language processing at the university.
Data science is one of the fastest growing fields in the world today.
The teacher asked the students to submit their assignments on time.
He was reading a book while sitting on the chair by the window.
Artificial intelligence and machine learning are transforming the industry.
The city was built along the river and near the mountains.
They were discussing the impact of technology on modern education.
A good solution to the problem requires careful analysis and planning.
I went to the market to buy fruits and vegetables for the week.
The results of the experiment were recorded in the final report.
This project is focused on improving accuracy and performance.
Many people believe that practice is the key to success in life.
The company announced an increase in profit for the next quarter.
'''
```

```python
import nltk
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
from nltk.tokenize import word_tokenize
nltk.ne_chunk(nltk.pos_tag(word_tokenize(corpus))).draw()
```

```python
ner_tree=nltk.pos_tag(word_tokenize(corpus))
print(ner_tree)
entities = []
for chunk in ner_tree:
    if hasattr(chunk, 'label'):
        entity = ' '.join(c[0] for c in chunk)
        entities.append((entity, chunk.label()))
print(entities)  # [('Bill Gates', 'PERSON'), ('Microsoft', 'ORGANIZATION'), ('Seattle', 'GPE')]


##  A named entity in NLP refers to a specific, real-world object or concept like a 
# person, organization, location, date, or product that carries key semantic meaning in text.
'''
ORGANIZATION - Georgia-Pacific Corp., WHO
PERSON - Eddy Bonte, President Obama
LOCATION - Murray River, Mount Everest
DATE - June, 2008-06-29
TIME - two fifty a m, 1:30 p.m.
MONEY - 175 million Canadian Dollars, GBP 10.40
PERCENT - twenty pct, 18.75 %
FACILITY - Washington Monument, Stonehenge
GPE - South East Asia, Midlothian
'''

## NO SPECIFIC ENTITY FOUND IN CORPUS 
```

```python
'''Named Entity Recognition (NER) identifies and classifies specific 
named entities like persons, organizations, locations, dates, and more
in text, enabling NLP systems to extract structured information for deeper
understanding.

Key Uses
Extracts real-world entities (e.g., "Apple" as organization vs. fruit) for
tasks like information retrieval, question answering, and knowledge graph building.
​

Powers applications such as search engines, chatbots, and sentiment analysis
by linking text to entities.
​

Facilitates data summarization and event detection in news or legal documents.'''
```

```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk

corpus = "Bill Gates founded Microsoft in Seattle."
tokens = word_tokenize(corpus)
pos_tags = pos_tag(tokens)
ner_tree = ne_chunk(pos_tags)
print(ner_tree)  # Shows tree

# Extract entities
entities = []
for chunk in ner_tree:
    if hasattr(chunk, 'label'):
        entity = ' '.join(c[0] for c in chunk) ## person, place , noun
        entities.append((entity, chunk.label())) ## chunk label gives enityt label 
print(entities)  # [('Bill Gates', 'PERSON'), ('Microsoft', 'ORGANIZATION'), ('Seattle', 'GPE')]
```
