```python

from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
corpus = []
for i in range(0, len(df)):
    review = re.sub('[^a-zA-Z]', ' ', df['text'][i])
    review = review.lower()
    review = review.split()
    
    review = [lemmatizer.lemmatize(word) for word in review]
    review = ' '.join(review)
    corpus.append(review)
    
#corpus ->[sent, sent , sent....]
```

```python
texts = df['text'].tolist()
y = df['label'].tolist()
```

```python
corpus = []
for text in texts:
    review = re.sub('[^a-zA-Z]', ' ', text)
    review = review.lower().split()
    review = [lemmatizer.lemmatize(word) for word in review]
    corpus.append(review)   # KEEP AS LIST, not string
```

```python
indexed_words = []

for i, doc in enumerate(corpus):
    if len(doc) > 0:
        indexed_words.append((i, simple_preprocess(" ".join(doc))))
```

```python
## do splitting first then vectorizing 

words_only = [w for _, w in indexed_words]
model1 = gensim.models.Word2Vec(words_only, vector_size=100)
```

```python
def avg_word2vec(doc):
    vectors = [model1.wv[word] for word in doc if word in model1.wv.index_to_key]
    if len(vectors) == 0:
        return None
    return np.mean(vectors, axis=0)
```

```python
X = []
y_clean = []
text_clean = []

for idx, doc in indexed_words:
    vec = avg_word2vec(doc)
    if vec is not None:
        X.append(vec)
        y_clean.append(y[idx])
        text_clean.append(texts[idx])
```

```python
assert len(X) == len(y_clean) == len(text_clean)
```

```python
df1 = pd.DataFrame(X)
```

```python
y_clean = pd.get_dummies(y_clean,drop_first=True,dtype=int)
```

```python
df1['output']=y_clean
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df1.drop(columns=['output']),df1['output'],test_size=0.20)
```
