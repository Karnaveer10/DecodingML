```python
corpus ='''The quick brown fox jumps over the lazy dog near the river bank.
In the morning, the weather was cold and the sky was filled with clouds.
She is studying natural language processing at the university.
Data science is one of the fastest growing fields in the world today.
The teacher asked the students to submit their assignments on time.
He was reading a book while sitting on the chair by the window.
Artificial intelligence and machine learning are transforming the industry.
The city was built along the river and near the mountains.
They were discussing the impact of technology on modern education.
A good solution to the problem requires careful analysis and planning.
I went to the market to buy fruits and vegetables for the week.
The results of the experiment were recorded in the final report.
This project is focused on improving accuracy and performance.
Many people believe that practice is the key to success in life.
The company announced an increase in profit for the next quarter.
'''
```

```python

# Stop-word removal is an important preprocessing step in Natural Language Processing (NLP)

# It helps reduce noise in text data by removing commonly occurring words 
# (such as 'the', 'is', 'and') that carry little meaningful information.

# Removing stop words improves computational efficiency by reducing
# the size of the vocabulary and the number of tokens to process.

# It increases the relevance of important words, making models focus
# on content-bearing terms rather than grammatical fillers.

# Stop-word removal enhances the performance of tasks like
# text classification, clustering, and topic modeling.

# It helps improve accuracy in information retrieval systems
# by prioritizing meaningful keywords during search.

# In sentiment analysis, removing stop words can highlight
# opinion-carrying words such as adjectives and verbs.

# In machine learning models like Bag-of-Words and TF-IDF,
# stop-word removal prevents high-frequency common words
# from dominating feature representations.

# Applications of stop-word removal include:
# - Search engines
# - Spam filtering
# - Text summarization
# - Chatbots and virtual assistants
# - Recommendation systems
# - Document indexing and retrieval
```

```python
from nltk.corpus import stopwords
```

```python
import nltk 
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
import nltk
from nltk.tokenize import sent_tokenize,word_tokenize

nltk.download("punkt")
nltk.download('punkt_tab')
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
lemm = WordNetLemmatizer()
```

```python
## Flow --> Applying Stop Words , Applying Lemmetization
import string
sentences = sent_tokenize(corpus)

for i,sent in enumerate(sentences):

    filtered = [
        word.lower()
        for word in word_tokenize(sent)
        if word.lower() not in set(stopwords.words('english'))
        and word.lower() not in set(string.punctuation)
    ]
    lemmetized = [lemm.lemmatize(word,pos='v') for word in filtered]
    sentences[i]=' '.join(lemmetized)
# for sent in sentences:
#     filtered = [word for word in sent if word not in stopwords.words('english')]
print(sentences)
```
